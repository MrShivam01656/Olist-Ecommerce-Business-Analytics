```python
# Check referential integrity between related datasets

checks = {
    "Orders → Customers": (
        orders_df["customer_id"],
        customers_df["customer_id"]
    ),

    "Order Items → Orders": (
        order_items_df["order_id"],
        orders_df["order_id"]
    ),

    "Order Items → Products": (
        order_items_df["product_id"],
        products_df["product_id"]
    ),

    "Order Items → Sellers": (
        order_items_df["seller_id"],
        sellers_df["seller_id"]
    ),

    "Payments → Orders": (
        payments_df["order_id"],
        orders_df["order_id"]
    ),

    "Reviews → Orders": (
        reviews_df["order_id"],
        orders_df["order_id"]
    )
}

for relationship, (child_ids, parent_ids) in checks.items():

    missing_ids = child_ids[~child_ids.isin(parent_ids)]

    print(f"{relationship}")
    print(f"Missing IDs: {missing_ids.nunique():,}")
    print("-" * 60)
```

### Order and Order Items Relationship

The Orders table contains one row per order, while the Order Items table contains one row per item within an order.

The Order Items dataset contains 112,650 records corresponding to 98,666 unique orders. A single order can contain multiple items, with a maximum of 21 items in one order.

Therefore, the relationship between Orders and Order Items is one-to-many.

```python
# ============================================================
# CREATE PROCESSED ANALYTICAL TABLES


from pathlib import Path
import pandas as pd

# Paths
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "Data" / "Raw"
PROCESSED_DATA_PATH = PROJECT_ROOT / "Data" / "Processed"

PROCESSED_DATA_PATH.mkdir(parents=True, exist_ok=True)

print("Processed folder:", PROCESSED_DATA_PATH)



# 1. LOAD REQUIRED RAW DATA
# ============================================================

orders_df = pd.read_csv(
    RAW_DATA_PATH / "olist_orders_dataset.csv"
)

customers_df = pd.read_csv(
    RAW_DATA_PATH / "olist_customers_dataset.csv"
)

order_items_df = pd.read_csv(
    RAW_DATA_PATH / "olist_order_items_dataset.csv"
)

products_df = pd.read_csv(
    RAW_DATA_PATH / "olist_products_dataset.csv"
)

translation_df = pd.read_csv(
    RAW_DATA_PATH / "product_category_name_translation.csv"
)

payments_df = pd.read_csv(
    RAW_DATA_PATH / "olist_order_payments_dataset.csv"
)

reviews_df = pd.read_csv(
    RAW_DATA_PATH / "olist_order_reviews_dataset.csv"
)



# 2. PAYMENT SUMMARY
# ============================================================

payment_summary = (
    payments_df
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_record_count=("payment_sequential", "count"),
        max_payment_installments=("payment_installments", "max")
    )
    .reset_index()
)

payment_methods = (
    payments_df
    .groupby("order_id")["payment_type"]
    .agg(lambda x: " + ".join(sorted(x.unique())))
    .reset_index(name="payment_methods")
)

payment_summary = payment_summary.merge(
    payment_methods,
    on="order_id",
    how="left"
)


# ============================================================
# 3. REVIEW SUMMARY


reviews_df["has_comment"] = (
    reviews_df["review_comment_message"].notna()
)

review_summary = (
    reviews_df
    .groupby("order_id")
    .agg(
        review_count=("review_id", "count"),
        average_review_score=("review_score", "mean"),
        minimum_review_score=("review_score", "min"),
        maximum_review_score=("review_score", "max"),
        comment_count=("has_comment", "sum"),
        has_comment=("has_comment", "any")
    )
    .reset_index()
)

review_summary["average_review_score"] = (
    review_summary["average_review_score"].round(2)
)


# ============================================================
# 4. ORDERS ANALYTICS


customer_info = customers_df[
    [
        "customer_id",
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ]
]

orders_analytics = (
    orders_df
    .merge(
        customer_info,
        on="customer_id",
        how="left"
    )
    .merge(
        payment_summary,
        on="order_id",
        how="left"
    )
    .merge(
        review_summary,
        on="order_id",
        how="left"
    )
)

# Convert dates
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_analytics[col] = pd.to_datetime(
        orders_analytics[col],
        errors="coerce"
    )

# Time features
orders_analytics["purchase_date"] = (
    orders_analytics["order_purchase_timestamp"].dt.date
)

orders_analytics["purchase_year"] = (
    orders_analytics["order_purchase_timestamp"].dt.year
)

orders_analytics["purchase_month"] = (
    orders_analytics["order_purchase_timestamp"].dt.month
)

orders_analytics["purchase_month_name"] = (
    orders_analytics["order_purchase_timestamp"].dt.strftime("%B")
)

orders_analytics["purchase_day_of_week"] = (
    orders_analytics["order_purchase_timestamp"].dt.day_name()
)

orders_analytics["purchase_hour"] = (
    orders_analytics["order_purchase_timestamp"].dt.hour
)

# Delivery metrics
orders_analytics["delivery_days"] = (
    orders_analytics["order_delivered_customer_date"]
    - orders_analytics["order_purchase_timestamp"]
).dt.total_seconds() / (24 * 60 * 60)

orders_analytics["delivery_delay_days"] = (
    orders_analytics["order_delivered_customer_date"].dt.normalize()
    - orders_analytics["order_estimated_delivery_date"].dt.normalize()
).dt.total_seconds() / (24 * 60 * 60)

orders_analytics["delivery_performance"] = "Not Delivered"

delivered_mask = (
    orders_analytics["order_status"] == "delivered"
)

orders_analytics.loc[
    delivered_mask &
    (orders_analytics["delivery_delay_days"] < 0),
    "delivery_performance"
] = "Early"

orders_analytics.loc[
    delivered_mask &
    (orders_analytics["delivery_delay_days"] == 0),
    "delivery_performance"
] = "On Time"

orders_analytics.loc[
    delivered_mask &
    (orders_analytics["delivery_delay_days"] > 0),
    "delivery_performance"
] = "Late"


# ============================================================
# 5. CUSTOMER SUMMARY


customer_summary = (
    orders_analytics
    .groupby("customer_unique_id")
    .agg(
        total_orders=("order_id", "nunique"),
        total_spent=("total_payment_value", "sum"),
        average_order_value=("total_payment_value", "mean")
    )
    .reset_index()
)

customer_summary["customer_type"] = (
    customer_summary["total_orders"]
    .apply(
        lambda x: "Repeat" if x > 1 else "One-Time"
    )
)


# ============================================================
# 6. PRODUCT CATEGORY MAPPING


manual_category_mapping = {
    "pc_gamer": "gaming_pc",
    "portateis_cozinha_e_preparadores_de_alimentos":
        "kitchen_food_preparation_appliances"
}

products_category = (
    products_df[
        ["product_id", "product_category_name"]
    ]
    .merge(
        translation_df,
        on="product_category_name",
        how="left"
    )
)

products_category[
    "product_category_name_english"
] = (
    products_category[
        "product_category_name_english"
    ]
    .fillna(
        products_category["product_category_name"]
        .map(manual_category_mapping)
    )
)


# ============================================================
# 7. ORDER ITEMS ANALYTICS


order_items_analytics = (
    order_items_df
    .merge(
        products_category[
            [
                "product_id",
                "product_category_name_english"
            ]
        ],
        on="product_id",
        how="left"
    )
)

order_items_analytics["item_total_value"] = (
    order_items_analytics["price"]
    + order_items_analytics["freight_value"]
)


# ============================================================
# 8. CATEGORY SUMMARY


category_summary = (
    order_items_analytics
    .groupby("product_category_name_english")
    .agg(
        items_sold=("product_id", "count"),
        orders=("order_id", "nunique"),
        product_sales=("price", "sum"),
        freight_revenue=("freight_value", "sum")
    )
    .reset_index()
)

category_summary["total_value"] = (
    category_summary["product_sales"]
    + category_summary["freight_revenue"]
)

category_summary = category_summary.sort_values(
    "product_sales",
    ascending=False
)


# ============================================================
# 9. SELLER SUMMARY


seller_summary = (
    order_items_df
    .groupby("seller_id")
    .agg(
        items_sold=("order_id", "count"),
        orders=("order_id", "nunique"),
        product_sales=("price", "sum"),
        freight_revenue=("freight_value", "sum")
    )
    .reset_index()
)

seller_summary["total_value"] = (
    seller_summary["product_sales"]
    + seller_summary["freight_revenue"]
)

seller_summary = seller_summary.sort_values(
    "product_sales",
    ascending=False
)


# ============================================================
# 10. MONTHLY SALES


monthly_sales = (
    orders_analytics
    .groupby(
        [
            "purchase_year",
            "purchase_month"
        ]
    )
    .agg(
        orders=("order_id", "nunique"),
        payment_value=("total_payment_value", "sum")
    )
    .reset_index()
)

monthly_sales["year_month"] = pd.to_datetime(
    monthly_sales["purchase_year"].astype(str)
    + "-"
    + monthly_sales["purchase_month"].astype(str)
    + "-01"
)

monthly_sales = monthly_sales.sort_values("year_month")


# ============================================================
# 11. STATE SALES


state_sales = (
    orders_analytics
    .groupby("customer_state")
    .agg(
        customers=("customer_unique_id", "nunique"),
        orders=("order_id", "nunique"),
        payment_value=("total_payment_value", "sum")
    )
    .reset_index()
)

state_sales["sales_share_pct"] = (
    state_sales["payment_value"]
    / state_sales["payment_value"].sum()
    * 100
)

state_sales["sales_share_pct"] = (
    state_sales["sales_share_pct"].round(2)
)

state_sales = state_sales.sort_values(
    "payment_value",
    ascending=False
)


# ============================================================
# 12. SAVE ALL PROCESSED TABLES


tables = {
    "orders_analytics": orders_analytics,
    "order_items_analytics": order_items_analytics,
    "customer_summary": customer_summary,
    "category_summary": category_summary,
    "seller_summary": seller_summary,
    "monthly_sales": monthly_sales,
    "state_sales": state_sales,
    "payment_summary": payment_summary,
    "review_summary": review_summary
}

for table_name, table_df in tables.items():
    output_path = (
        PROCESSED_DATA_PATH / f"{table_name}.csv"
    )

    table_df.to_csv(
        output_path,
        index=False
    )

    print(
        f"Saved: {table_name}.csv | "
        f"Rows: {len(table_df):,} | "
        f"Columns: {len(table_df.columns)}"
    )

print("\n" + "=" * 60)
print("ALL PROCESSED ANALYTICAL TABLES CREATED SUCCESSFULLY")
print("=" * 60)
```
